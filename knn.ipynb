```python
import numpy as np
import matplotlib.pyplot as plt
import random
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
```

```python
X = np.random.rand(100,2)
y = np.random.rand(100,1)
for i in range(100):
    if y[i] < 0.5:
        y[i] = 0
    else:
        y[i] = 1
```

```python
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
accuracy_train = {}
accuracy_test = {}  
neighbors = np.arange(1,30)

for neighbor in neighbors:
    model = KNeighborsClassifier(n_neighbors=neighbor)
    model.fit(x_train, y_train)
    accuracy_train[neighbor] = model.score(x_train, y_train)
    accuracy_test[neighbor] = model.score(x_test, y_test)

plt.figure(figsize=(10,6))
plt.plot(accuracy_train.keys(), accuracy_train.values(), label='Train Accuracy') 
plt.plot(accuracy_test.keys(), accuracy_test.values(), label='Test Accuracy') 
plt.xlabel('Number of Neighbors')
plt.ylabel('Accuracy')   

plt.show() 
```
